```python
#checking negstive value
numeric_cols = [
    'impressions',
    'clicks',
    'ad_spend',
    'conversions',
    'revenue'
]

for col in numeric_cols:
    print(col, (df[col] < 0).sum())
```

```python
#check whether clicks exceed impressions
(df['clicks']>df['impressions']).sum()
```

```python
#check whether conversions exceed clicks
(df['conversions']>df['clicks']).sum()
```

```python
df['calculated_CTR'] = (
    df['clicks'] / df['impressions']
) 

df['calculated_CPC'] = (
    df['ad_spend'] / df['clicks']
)

df['calculated_CPA'] = (
    df['ad_spend'] / df['conversions']
)

df['calculated_ROAS'] = (
    df['revenue'] / df['ad_spend']
)

#Profit Calculation
df["profit"] = df["revenue"] - df["ad_spend"]

#ROI percentage
df['ROI percent']=((df['revenue']-df['ad_spend'])/df['ad_spend'])*100
```

```python
df.head()
```

```python
#Compare the original and calculated metrics
df[[
    'CTR', 'calculated_CTR',
    'CPC', 'calculated_CPC',
    'CPA', 'calculated_CPA',
    'ROAS', 'calculated_ROAS'
]].head()
```

```python
final_cols=[
    'date', 'platform', 
    'campaign_type', 'industry', 
    'country','impressions','clicks',
    'ad_spend', 'conversions','revenue',
    'calculated_CTR', 'calculated_CPC', 
    'calculated_CPA','calculated_ROAS', 
    'ROI percent', 'profit'    
]

df_clean = df[final_cols]
```

```python
df_clean.head()
```

```python
# Export cleaned dataset
df_clean.to_csv(
    "marketing_campaign_cleaned.csv",
    index=False
)

print("✅ Clean dataset saved successfully")
```
